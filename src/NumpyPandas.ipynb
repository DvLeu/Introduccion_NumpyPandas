**Tipos de Datos**
| dtype              | Rango                | Bits    |
| ------------------ | -------------------- | ------- |
| `int8` / `uint8`   | -128 a 127 / 0 a 255 | 8 bits  |
| `int16` / `uint16` | ±32,767              | 16 bits |
| `int32` / `uint32` | ±2e9                 | 32 bits |
| `int64` / `uint64` | ±9e18                | 64 bits |

---

| dtype     | Precisión | Notas                    |
| --------- | --------- | ------------------------ |
| `float16` | baja      | computación rápida (GPU) |
| `float32` | media     | machine learning         |
| `float64` | alta      | por defecto en NumPy     |
