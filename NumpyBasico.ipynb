# Generación de números Aleatorios
- **rand()**, crea un arreglo de números aleatorios con una distribución uniforme entre 0 y 1.
- **randn()**, Crea un arreglo de números aleatorios con una distribución normal estándar.
- **randint()**, crea un arreglo de números aleatorios enteros en un rango dado.

```python
import numpy as np

#dir(np.random)
#help(np.random.randint)

arreglo_uniforme=np.random.rand(3,3)
arreglo_estandar=np.random.randn(4,4)
arreglo_enteros=np.random.randint(low=10,high=50,size=(3,3)) # si omito el high

print(arreglo_uniforme)
print(arreglo_estandar)
print(arreglo_enteros)
```

```python
# normal(media=0, std=1, tamaño=n)

a = np.random.normal(size=10)
a
```

```python
b= np.random.normal(10,2,(5,5))
b
```

```python
np.mean(b)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.random.normal(loc=0, scale=1, size=1000)
media=np.mean(data)
devstd=np.std(data)
plt.scatter(data,norm.pdf(data,media,devstd))

plt.title("Distribución Normal")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

```python
# Binomial(numero_intentos, probabilidad_exito, tamaño)

b = np.random.binomial(10,0.5,(5,20))
b
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5
size = 100

b = np.random.binomial(n, p, size)

plt.hist(b, bins=np.arange(-0.5, n+1.5, 1), density=True, edgecolor='black', alpha=0.7, label='Histogram')

x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

plt.scatter(x, pmf, color='red', label='Theoretical PMF')
plt.vlines(x, 0, pmf, colors='red', linestyles='dashed')

plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Poisson (lambda, size)

import numpy as np
m = np.random.poisson(lam=4, size=(2, 3))
print(m)
```

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 3
data = np.random.poisson(lam=lam, size=1000)

plt.hist(data, bins=np.arange(-0.5, max(data)+1.5, 1), edgecolor='black', density=True)
plt.title(f"Distribución Poisson (λ={lam})")
plt.xlabel("Numero de eventos")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
# Uniforme

import numpy as np
arr = np.random.uniform(size=5)
print(arr)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

low = 10
high = 20
size = 1000

data = np.random.uniform(low, high, size)

sns.histplot(data, bins=30, kde=False, color='skyblue', edgecolor='black')

plt.title(f"Distribución Uniforme (Rango: {low} a {high})")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.grid(True)
plt.show()
```

```python
# Exponencial

import numpy as np
arr = np.random.exponential(scale=1.5, size=5)
print(arr)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

s = 2      # scale
n = 800    # number of points

data = np.random.exponential(scale=s, size=n)
sns.histplot(data, bins=30, kde=True, edgecolor='black')

plt.title(f"Distribución Exponencial (Scale={s})")
plt.xlabel("Valor")
plt.ylabel("Frecuencía")
plt.grid(True)
plt.show()
```

```python
# Chi-cuadrada (grados de libertad y tamaño)

import numpy as np
arr = np.random.chisquare(df=2, size=5)
print(arr)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

df = 2
size = 1000

data = np.random.chisquare(df, size)
plt.hist(data, bins=30, density=True, edgecolor='black', alpha=0.7, label='Histogram')

x = np.linspace(0, max(data), 200)
pdf = chi2.pdf(x, df)
plt.plot(x, pdf, color='red', label='Theoretical PDF')

plt.title(f"Distribución Chi-Square (df={df})")
plt.xlabel("Valores")
plt.ylabel("Densidad")
plt.legend()
plt.grid(True)
plt.show()
```
